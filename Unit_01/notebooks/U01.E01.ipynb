```python
# =============================================================================
# ACTIVIDAD 4
# Estadística descriptiva de las ventas mensuales
# =============================================================================

import pandas as pd
import numpy as np

# -----------------------------------------------------------------------------
# Datos de ventas mensuales (miles de RD$)
# -----------------------------------------------------------------------------

ventas = [100, 120, 120, 140, 160, 260]

# -----------------------------------------------------------------------------
# Cálculo de estadísticas descriptivas
# -----------------------------------------------------------------------------

media = np.mean(ventas)
mediana = np.median(ventas)
moda = pd.Series(ventas).mode()[0]
minimo = min(ventas)
maximo = max(ventas)
rango = maximo - minimo

# -----------------------------------------------------------------------------
# Mostrar resultados
# -----------------------------------------------------------------------------

print(f"Media: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Moda: {moda}")
print(f"Valor mínimo: {minimo}")
print(f"Valor máximo: {maximo}")
print(f"Rango: {rango}")
```

# Actividad 5
## Interpretación de la media

La media aritmética de las ventas es:

**150 miles de RD**

La mediana es:

**130 miles de RD**

La media es mayor que la mediana debido al valor observado en junio (**260 miles de RD**), que es considerablemente superior al resto de las observaciones.

Este comportamiento sugiere que junio ejerce una influencia importante sobre el promedio general, elevándolo por encima del nivel habitual de ventas observado durante la mayor parte del semestre.

Por esta razón, la media representa parcialmente el comportamiento típico de las ventas. Aunque es una medida válida del promedio general, la mediana proporciona una descripción más representativa de una venta mensual típica porque es menos sensible a valores extremos.

### Conclusión

El valor de junio incrementa la media de las ventas. Para describir el comportamiento habitual del período, es recomendable analizar conjuntamente la media y la mediana.

```python
# =============================================================================
# ACTIVIDAD 6
# Construcción de los conjuntos de datos
# =============================================================================

import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------------------------------------------------------
# Datos de ventas y reclamaciones mensuales
# -----------------------------------------------------------------------------

df_mensual = pd.DataFrame({
    "Mes": [
        "Enero",
        "Febrero",
        "Marzo",
        "Abril",
        "Mayo",
        "Junio"
    ],
    "Ventas": [100, 120, 120, 140, 160, 260],
    "Reclamaciones": [8, 9, 9, 12, 15, 25]
})

# -----------------------------------------------------------------------------
# Datos de ventas por canal
# -----------------------------------------------------------------------------

df_canal = pd.DataFrame({
    "Canal": [
        "Página web",
        "Aplicación móvil",
        "Redes sociales",
        "Ventas telefónicas"
    ],
    "Ventas": [100, 80, 50, 30]
})

# Mostrar los datos utilizados
display(df_mensual)
display(df_canal)
```

### Gráfico 1. Evolución de las ventas mensuales

Se utiliza un gráfico de líneas porque la variable "Mes" tiene una dimensión temporal y este tipo de gráfico facilita la visualización de tendencias a lo largo del tiempo.

```python
# =============================================================================
# Gráfico de líneas
# Evolución de las ventas mensuales
# =============================================================================

plt.figure(figsize=(8, 5))

plt.plot(
    df_mensual["Mes"],
    df_mensual["Ventas"],
    marker="o",
    linewidth=2
)

plt.title("Evolución de las ventas mensuales")
plt.xlabel("Mes")
plt.ylabel("Ventas (miles de RD)")
plt.grid(True)

plt.figtext(
    0.5,
    -0.05,
    "Fuente: ComercioDigital RD",
    ha="center"
)

plt.show()
```

### Gráfico 2. Comparación de ventas por canal

Se utiliza un gráfico de barras porque permite comparar fácilmente valores entre categorías independientes.

```python
# =============================================================================
# Gráfico de barras
# Ventas por canal en junio
# =============================================================================

plt.figure(figsize=(8, 5))

plt.bar(
    df_canal["Canal"],
    df_canal["Ventas"],
    color="steelblue"
)

plt.title("Ventas por canal - Junio")
plt.xlabel("Canal")
plt.ylabel("Ventas (miles de RD)")

plt.figtext(
    0.5,
    -0.05,
    "Fuente: ComercioDigital RD",
    ha="center"
)

plt.show()
```

### Gráfico 3. Participación de ventas por canal

Se utiliza un gráfico circular para visualizar qué proporción del total de ventas corresponde a cada canal comercial.

```python
# =============================================================================
# Gráfico circular
# Participación de ventas por canal
# =============================================================================

plt.figure(figsize=(7, 7))

plt.pie(
    df_canal["Ventas"],
    labels=df_canal["Canal"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Participación de ventas por canal - Junio")

plt.show()
```

# Actividad 8

## Relación entre ventas y reclamaciones

Para explorar la relación entre las ventas y las reclamaciones mensuales se utilizaría un **gráfico de dispersión (Scatter Plot)**.

En este gráfico, las ventas se representarían en el eje X y las reclamaciones en el eje Y. Cada punto correspondería a un mes del período analizado.

### ¿Por qué utilizar este gráfico?

El gráfico de dispersión es adecuado porque permite visualizar si existe una asociación entre dos variables cuantitativas. Además, facilita la identificación de patrones, tendencias, agrupamientos o posibles valores atípicos.

### Interpretación preliminar

Al observar los datos, se aprecia que las reclamaciones aumentan a medida que aumentan las ventas. Esto sugiere la existencia de una posible relación positiva entre ambas variables.

### Precaución al interpretar los resultados

Aunque las ventas y las reclamaciones parecen aumentar conjuntamente, esto no implica que una variable sea la causa de la otra. La correlación no demuestra causalidad.

Es posible que ambas variables aumenten debido a un tercer factor, como el incremento del número de clientes, campañas promocionales, expansión del negocio o un mayor volumen de transacciones.

Por lo tanto, antes de concluir que las reclamaciones provocan cambios en las ventas o viceversa, sería necesario realizar estudios adicionales y disponer de más información que permita evaluar relaciones causales.
