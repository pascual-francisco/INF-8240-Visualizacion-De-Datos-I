```python
# ------------------------------------------------------------------------------
# Synchronize current notebook with repository
# ------------------------------------------------------------------------------

from google.colab import _message
from pathlib import Path
import json

# ------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------

UNIT = "Unit_01"
EXERCISE = "U01.E01"

# ------------------------------------------------------------------------------
# DESTINATION PATH
# ------------------------------------------------------------------------------

target = Path(
    "/content/INF-8240-Visualizacion-De-Datos-I"
) / UNIT / "notebooks" / f"{EXERCISE}.ipynb"

# ------------------------------------------------------------------------------
# VALIDATE DIRECTORY
# ------------------------------------------------------------------------------

target.parent.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------------------------
# RETRIEVE CURRENT NOTEBOOK
# ------------------------------------------------------------------------------

nb = _message.blocking_request(
    "get_ipynb"
)

# ------------------------------------------------------------------------------
# SAVE NOTEBOOK
# ------------------------------------------------------------------------------

with target.open(
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        nb["ipynb"],
        file,
        ensure_ascii=False,
        indent=1
    )

print("✅ Notebook saved successfully:")
print(target)
```

```python
# ==============================================================================
# AUTOMATIC COMMIT AND AUTHENTICATED PUSH
# ==============================================================================

import os
import stat
import subprocess
from pathlib import Path
from google.colab import userdata

# ------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------

REPO_PATH = Path(
    "/content/INF-8240-Visualizacion-De-Datos-I"
)

GITHUB_USER = "pascual-francisco"
GITHUB_REPO = "INF-8240-Visualizacion-De-Datos-I"

COMMIT_MESSAGE = "chore: sync repository updates"

GITHUB_TOKEN = userdata.get("TOKEN")

# ------------------------------------------------------------------------------

def run_git(
    arguments,
    *,
    check=True,
    capture_output=False,
    environment=None
):
    return subprocess.run(
        ["git", *arguments],
        cwd=REPO_PATH,
        check=check,
        text=True,
        capture_output=capture_output,
        env=environment
    )

# ------------------------------------------------------------------------------
# VALIDATE TOKEN
# ------------------------------------------------------------------------------

if not GITHUB_TOKEN:
    raise ValueError(
        "TOKEN not found in Colab Secrets."
    )

# ------------------------------------------------------------------------------
# STAGE CHANGES
# ------------------------------------------------------------------------------

print("=" * 70)
print("STAGING CHANGES")
print("=" * 70)

run_git(["add", "-A"])

# ------------------------------------------------------------------------------
# CREATE COMMIT IF NEEDED
# ------------------------------------------------------------------------------

staged = run_git(
    ["diff", "--cached", "--name-status"],
    capture_output=True
)

staged_output = staged.stdout.strip()

if staged_output:

    print(staged_output)

    run_git(
        [
            "commit",
            "-m",
            COMMIT_MESSAGE
        ]
    )

    print(
        f"\nCommit created: {COMMIT_MESSAGE}"
    )

else:

    print(
        "\nNo new changes to commit."
    )

# ------------------------------------------------------------------------------
# CURRENT BRANCH
# ------------------------------------------------------------------------------

branch = run_git(
    [
        "branch",
        "--show-current"
    ],
    capture_output=True
).stdout.strip()

print(f"\nCurrent branch: {branch}")

# ----
```
