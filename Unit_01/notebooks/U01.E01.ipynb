```python
# ------------------------------------------------------------------------------
# Synchronize current notebook with repository
# ------------------------------------------------------------------------------

from google.colab import _message
from pathlib import Path
import json

# ------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------

UNIT = "Unit_01"
EXERCISE = "U01.E01"

# ------------------------------------------------------------------------------
# DESTINATION PATH
# ------------------------------------------------------------------------------

target = Path(
    "/content/INF-8240-Visualizacion-De-Datos-I"
) / UNIT / "notebooks" / f"{EXERCISE}.ipynb"

# ------------------------------------------------------------------------------
# VALIDATE DIRECTORY
# ------------------------------------------------------------------------------

target.parent.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------------------------
# RETRIEVE CURRENT NOTEBOOK
# ------------------------------------------------------------------------------

nb = _message.blocking_request(
    "get_ipynb"
)

# ------------------------------------------------------------------------------
# SAVE NOTEBOOK
# ------------------------------------------------------------------------------

with target.open(
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        nb["ipynb"],
        file,
        ensure_ascii=False,
        indent=1
    )

print("✅ Notebook saved successfully:")
print(target)
```
