```python
# ------------------------------------------------------------------------------
# Synchronize current notebook with repository
# ------------------------------------------------------------------------------

from google.colab import _message
from pathlib import Path
import json

# ------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------

UNIT = "Unit_01"
EXERCISE = "U01.E01"

# ------------------------------------------------------------------------------
# DESTINATION PATH
# ------------------------------------------------------------------------------

target = Path(
    "/content/INF-8240-Visualizacion-De-Datos-I"
) / UNIT / "notebooks" / f"{EXERCISE}.ipynb"

# ------------------------------------------------------------------------------
# VALIDATE DIRECTORY
# ------------------------------------------------------------------------------

target.parent.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------------------------
# RETRIEVE CURRENT NOTEBOOK
# ------------------------------------------------------------------------------

nb = _message.blocking_request(
    "get_ipynb"
)

# ------------------------------------------------------------------------------
# SAVE NOTEBOOK
# ------------------------------------------------------------------------------

with target.open(
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        nb["ipynb"],
        file,
        ensure_ascii=False,
        indent=1
    )

print("✅ Notebook saved successfully:")
print(target)
```

```python
# ==============================================================================
# COMMIT AND PUSH TO GITHUB
# ==============================================================================

from google.colab import userdata
from pathlib import Path
import subprocess

# ------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------

REPO_PATH = Path(
    "/content/INF-8240-Visualizacion-De-Datos-I"
)

GITHUB_USER = "pascual-francisco"
GITHUB_REPO = "INF-8240-Visualizacion-De-Datos-I"

COMMIT_MESSAGE = "chore: sync repository updates"

TOKEN = userdata.get("TOKEN")

# ------------------------------------------------------------------------------
# VALIDATE TOKEN
# ------------------------------------------------------------------------------

if not TOKEN:
    raise ValueError(
        "TOKEN not found in Colab Secrets."
    )

# ------------------------------------------------------------------------------
# MOVE TO REPOSITORY
# ------------------------------------------------------------------------------

%cd {REPO_PATH}

# ------------------------------------------------------------------------------
# STAGE CHANGES
# ------------------------------------------------------------------------------

print("=" * 70)
print("STAGING CHANGES")
print("=" * 70)

!git add -A

# ------------------------------------------------------------------------------
# CHECK IF THERE ARE CHANGES
# ------------------------------------------------------------------------------

status = subprocess.run(
    ["git", "diff", "--cached", "--quiet"]
)

if status.returncode == 0:

    print("\nNo changes to commit.")

else:

    !git commit -m "{COMMIT_MESSAGE}"

# ------------------------------------------------------------------------------
# PUSH
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("PUSHING TO GITHUB")
print("=" * 70)

!git push https://{GITHUB_USER}:{TOKEN}@github.com/{GITHUB_USER}/{GITHUB_REPO}.git main

# ------------------------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL STATUS")
print("=" * 70)

!git status --short --branch
```
